```python
import pandas as pd
import numpy as np

# Read Titanic dataset
df = pd.read_csv("titanic.csv")

print("Titanic Dataset:")
print(df.head())
```

```python
print("Shape of Dataset:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()
```

```python
print("Statistical Summary:")
print(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Fill missing Age values with median
df['Age'] = df['Age'].fillna(df['Age'].median())

# Fill missing Embarked values with mode
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Drop Cabin because it contains many missing values
df = df.drop('Cabin', axis=1)

print("Missing Values After Cleaning:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
df = df.rename(columns={
    'PassengerId': 'Passenger_ID',
    'Pclass': 'Passenger_Class',
    'SibSp': 'Siblings_Spouses',
    'Parch': 'Parents_Children'
})

print("Renamed Columns:")
print(df.columns.tolist())
```

```python
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[0, 12, 18, 35, 60, 100],
    labels=['Child', 'Teenager', 'Young Adult', 'Adult', 'Senior']
)

print(df[['Age', 'Age_Group']].head(10))
```

```python
survived = df[df['Survived'] == 1]

print("Survived Passengers:")
print(survived.head())
```

```python
first_class = df[df['Passenger_Class'] == 1]

print("First Class Passengers:")
print(first_class.head())
```

```python
sorted_df = df.sort_values(by='Age', ascending=False)

print("Passengers sorted by Age:")
print(sorted_df[['Name', 'Age', 'Passenger_Class']].head(10))
```

```python
class_count = df.groupby('Passenger_Class')['Passenger_ID'].count()

print("Number of passengers in each class:")
print(class_count)
```

```python
survival_rate = df.groupby('Passenger_Class')['Survived'].mean()

print("Survival Rate by Passenger Class:")
print(survival_rate)
```

```python
df['Sex_Code'] = df['Sex'].map({
    'male': 0,
    'female': 1
})

print("Gender Transformation:")
print(df[['Sex', 'Sex_Code']].head())
```

```python
# DataFrame 1
df1 = df[['Passenger_ID', 'Name', 'Sex']].head(5)

# DataFrame 2
df2 = df[['Passenger_ID', 'Age', 'Fare']].head(5)

print("DataFrame 1:")
print(df1)

print("\nDataFrame 2:")
print(df2)
```

```python
merged_df = pd.merge(
    df1,
    df2,
    on='Passenger_ID'
)

print("After MERGE:")
print(merged_df)
```

```python
df3 = df[['Passenger_ID', 'Name', 'Sex']].head(3)

df4 = df[['Passenger_ID', 'Name', 'Sex']].iloc[3:6]

print("DataFrame 3:")
print(df3)

print("\nDataFrame 4:")
print(df4)
```

```python
concat_df = pd.concat(
    [df3, df4],
    ignore_index=True
)

print("After CONCAT:")
print(concat_df)
```

```python
# Set Passenger_ID as index
left_df = df[['Passenger_ID', 'Name']].head(5).set_index('Passenger_ID')

right_df = df[['Passenger_ID', 'Fare']].head(5).set_index('Passenger_ID')

print("Left DataFrame:")
print(left_df)

print("\nRight DataFrame:")
print(right_df)
```

```python
joined_df = left_df.join(right_df)

print("After JOIN:")
print(joined_df)
```

```python
print("Final Cleaned Dataset:")
print(df.head())

print("\nFinal Shape:")
print(df.shape)

print("\nRemaining Missing Values:")
print(df.isnull().sum())
```

```python
df.to_csv("Titanic_Cleaned.csv", index=False)

print("Titanic_Cleaned.csv saved successfully!")
```
